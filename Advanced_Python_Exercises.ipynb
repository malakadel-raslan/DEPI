## 9. Inheritance — Task
Create GraduateStudent.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")


class GraduateStudent(Student):
    def __init__(self, name, age, degree):
        super().__init__(name, age)
        self.degree = degree

    def display_info(self):
        super().display_info()
        print(f"Degree: {self.degree}")


grad = GraduateStudent("Malak", 22, "Master's")
grad.display_info()
```

## 10. Decorators — Task
Logger decorator.

```python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function: {func.__name__}")
        result = func(*args, **kwargs)
        print("Function finished")
        return result
    return wrapper


@logger
def greet(name):
    print(f"Hello, {name}!")


greet("Malak")
```

## 11. Generators — Task
Countdown generator.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1


for number in countdown(5):
    print(number)
```

## 12. Lambda, Map, Filter, Reduce — Task
Work with numbers list.

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Lambda: square a number
square = lambda x: x ** 2
print("Square of 4:", square(4))

# Map: square all numbers
squares = list(map(lambda x: x ** 2, numbers))
print("Squares:", squares)

# Filter: keep even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Even Numbers:", evens)

# Reduce: sum all numbers
total = reduce(lambda x, y: x + y, numbers)
print("Sum:", total)
```

## 13. Advanced Practice
- Max/min/avg function
- Agent class
- Generator divisible by 3

```python
def statistics(numbers):
    maximum = max(numbers)
    minimum = min(numbers)
    average = sum(numbers) / len(numbers)

    return maximum, minimum, average


nums = [10, 20, 30, 40, 50]

max_val, min_val, avg_val = statistics(nums)

print("Max:", max_val)
print("Min:", min_val)
print("Average:", avg_val)
```

```python
class Agent:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"Hello, I am {self.name}, your AI agent.")

    def perform_task(self, task):
        print(f"{self.name} is performing: {task}")


agent = Agent("Assistant")

agent.introduce()
agent.perform_task("Data Analysis")
```

```python
def divisible_by_three(limit):
    for num in range(limit + 1):
        if num % 3 == 0:
            yield num


for number in divisible_by_three(20):
    print(number)
```
